```python
import random

SURNAME = "Neri"
SEED_NUM = 9

# Seed initialization
random.seed(sum(ord(c) for c in SURNAME) + SEED_NUM)

# 1. Generate Sensor Data
raw_readings = []
for _ in range(10):
    if random.random() < 0.3:
        raw_readings.append(
            random.choice(["ERR", "N/A", "INVALID", "-5.2", "105.0"])
        )
    else:
        raw_readings.append(str(round(random.uniform(10, 95), 1)))

# 2. Process and Validate
valid_readings = []
invalid_count = 0

for entry in raw_readings:
    try:
        val = float(entry)
        if 0.0 <= val <= 100.0:
            valid_readings.append(val)
        else:
            invalid_count += 1
    except ValueError:
        invalid_count += 1

# 3. Classify Valid Readings
classification = {"Low (<30)": 0, "Normal (30-70)": 0, "High (>70)": 0}
for val in valid_readings:
    if val < 30.0:
        classification["Low (<30)"] += 1
    elif val <= 70.0:
        classification["Normal (30-70)"] += 1
    else:
        classification["High (>70)"] += 1

print("Generated Sensor Data:", raw_readings)
print(
    f"Valid/Invalid Results: Valid = {len(valid_readings)}, Invalid = {invalid_count}"
)
print("Classification Results:", classification)
print(
    f"Execution Log: Processed {len(raw_readings)} entries using seed ({SURNAME}_{SEED_NUM})."
)
print(
    f"Final Output: Valid Mean = {round(sum(valid_readings)/len(valid_readings), 2) if valid_readings else 0}"
)
```

```python
SURNAME = "Neri"
SEED_NUM = 9

# 1. Generate Signal
generated_signal = f"{SURNAME}#{2017 + SEED_NUM}!Sig_{SEED_NUM}"
processed_signal = generated_signal.strip()

# 2. Character-by-Character Analysis
letters = 0
digits = 0
spaces = 0
special = 0

for char in processed_signal:
    if char.isalpha():
        letters += 1
    elif char.isdigit():
        digits += 1
    elif char.isspace():
        spaces += 1
    else:
        special += 1

# 3. Signal Classification
signal_class = (
    "Alphanumeric Heavy"
    if (letters + digits) > (spaces + special)
    else "Mixed/Noise Heavy"
)

# Print Assessment Data
print("--- EXERCISE 2 RESULTS ---")
print("Generated Signal:", generated_signal)
print("Processed Signal:", processed_signal)
print(
    f"Character Analysis: Letters={letters}, Digits={digits}, Spaces={spaces}, Special={special}"
)
print("Signal Classification:", signal_class)
print(
    f"Execution Log: Analyzed {len(processed_signal)} characters successfully."
)
print(
    f"Final Output: Signal Type: [{signal_class}] | Length: {len(processed_signal)}"
)
```

```python
SURNAME = "Neri"
SEED_NUM = 9

# 1. Generate Password & Attempt Limit
generated_password = f"{SURNAME}#{SEED_NUM * 11}"
attempt_limit = 3

# Simulated failed password attempts
simulated_attempts = ["Neri123", "Neri#90", "pass1234"]

attempts_made = 0
access_result = "DENIED"
final_state = "LOCKED"

# 2. Authentication Monitor Process
for attempt in simulated_attempts:
    attempts_made += 1
    if attempt == generated_password:
        access_result = "GRANTED"
        final_state = "UNLOCKED"
        break
    if attempts_made >= attempt_limit:
        final_state = "TERMINATED / LOCKED OUT"
        break

# Print Assessment Data
print("--- EXERCISE 3 RESULTS ---")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)
print("Access Result:", access_result)
print("Final System State:", final_state)
print(f"Execution Log: Authentication ended after {attempts_made} attempt(s).")
print(f"Final Output: [{final_state}] - Access {access_result}")
```
