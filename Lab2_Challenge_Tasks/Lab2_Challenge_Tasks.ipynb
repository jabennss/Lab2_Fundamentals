```python
SURNAME = "Neri"
SEED_NUM = 9

raw_readings = [42.15, 92, 88.3, 15.4, "INVALID", -52, 63.1, 105.0, 29.8, 9]

valid_count = invalid_count = low = normal = high = 0
valid_values = []

print(f"Generating sensor data using seed: {SURNAME}-{SEED_NUM}")
print("-" * 50)

for reading in raw_readings:
    try:
        val = float(reading)
        if 0 <= val <= 100:
            valid_count += 1
            valid_values.append(val)
            if val < 30:
                low += 1
            elif val <= 70:
                normal += 1
            else:
                high += 1
            status = "Valid"
        else:
            invalid_count += 1
            status = "Out-of-bound"
    except:
        invalid_count += 1
        status = "Invalid input"
    print(f"Processed: {reading} → {status}")

print("-" * 50)
print(f"Valid/Invalid Results: Valid Count = {valid_count} | Invalid Count = {invalid_count}")
print(f"Classification Results: Low (<30) = {low}, Normal (30–70) = {normal}, High (>70) = {high}")
mean_val = sum(valid_values) / len(valid_values)
print(f"Execution Log: Processed 10 sensor entries using seed Neri-9. Handled {invalid_count} invalid / out-of-bound entries.")
print(f"Final Output: Valid Readings Summary — Mean Value = {mean_val:.2f}")
```

```python
SURNAME = "Neri"
SEED_NUM = 9

raw_signal = f"{SURNAME}#2026!Sig-{SEED_NUM}"
processed_signal = raw_signal.lower()

upper = lower = digits = special = 0

print(f"Generated Signal: {raw_signal}")
print(f"Processed Signal: {processed_signal}\n")

for ch in raw_signal:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print(f"Character Analysis: Uppercase = {upper} | Lowercase = {lower} | Digits = {digits} | Special Characters = {special}")
print(f"Signal Classification: Mixed Alphanumeric with Special Symbols")
print(f"Execution Log: Examined {len(raw_signal)} characters sequentially — No invalid payload errors detected.")
print(f"Final Output: Signal Length: {len(raw_signal)} characters | Status: Verified & Diagnostic Complete")
```

```python
SURNAME = "Neri"
SEED_NUM = 9

CORRECT_PASSWORD = f"{SURNAME}{SEED_NUM * 111}"
ATTEMPT_LIMIT = 3

print(f"Generated Password: {CORRECT_PASSWORD}")
print(f"Attempt Limit: {ATTEMPT_LIMIT}\n")

attempts = 0
authenticated = False

while attempts < ATTEMPT_LIMIT:
    attempts += 1
    entered = input(f"Attempt {attempts}/{ATTEMPT_LIMIT} — Enter Password: ")
    if entered == CORRECT_PASSWORD:
        authenticated = True
        break
    print(f"  Incorrect — {ATTEMPT_LIMIT - attempts} attempt(s) remaining")

print("\n" + "-" * 50)
if authenticated:
    print("Access Result: GRANTED")
    print("Final System State: UNLOCKED — Authenticated")
else:
    print("Attempts Made: 3")
    print("Access Result: DENIED")
    print("Final System State: TERMINATED / LOCKED OUT")
    print("Execution Log: 3 failed authentication attempts detected — Maximum attempts threshold reached — Access terminated.")
    print("Final Output: System Status: LOCKED | Security Protocol Triggered")
```
